```python
import cv2
import numpy as np
import insightface
from insightface.app import FaceAnalysis
from insightface.data import get_image as ins_get_image
from insightface.utils import face_align
```

```python
amin1 = "./amin1.jpg"
amin2 = "./amin2.jpg"
```

```python

# Initialize face analysis model
app = FaceAnalysis(name='buffalo_l', providers=['CPUExecutionProvider', 'CUDAExecutionProvider'])  # Use 'CUDAExecutionProvider' for GPU
app.prepare(ctx_id=0)  # ctx_id=-1 for CPU, 0 for GPU
```

```python
def get_face(image):
    """Extract face embedding from an image"""
    # if img is None:
    #     raise ValueError(f"Could not read image: {image_path}")
    
    faces = app.get(image)
    
    if len(faces) < 1:
        raise ValueError("No faces detected in the image")
    if len(faces) > 1:
        print("Warning: Multiple faces detected. Using first detected face")
    
    return faces[0]
```

```python
def compare_faces(emb1, emb2, threshold=0.65): # Adjust this threshold according to your usecase.
    """Compare two embeddings using cosine similarity"""
    similarity = np.dot(emb1, emb2) / (np.linalg.norm(emb1) * np.linalg.norm(emb2))
    
    return similarity, similarity > threshold
```

```python
def get_eye_color(img, eye_center):
    x, y = eye_center
    r = 10  # radius of patch around eye center
    eye_patch = img[y-r:y+r, x-r:x+r]
    
    # Convert to HSV
    hsv = cv2.cvtColor(eye_patch, cv2.COLOR_BGR2HSV)
    hue = hsv[:,:,0].flatten()

    # Compute dominant hue
    dominant_hue = np.median(hue)
    
    # Map hue to color
    if dominant_hue < 15 or dominant_hue > 160:
        return "Brown"
    elif dominant_hue < 35:
        return "Hazel"
    elif dominant_hue < 75:
        return "Green"
    elif dominant_hue < 135:
        return "Blue"
    else:
        return "Gray"
```

```python
try:
    # Get embeddings
    img1 = cv2.imread(amin1)
    face1 = get_face(img1)
    kps1 = face1.kps.astype(int)
    emb1 = face1.embedding

    img2 = cv2.imread(amin2)
    face2 = get_face(img2)
    kps2 = face2.kps.astype(int)
    emb2 = face2.embedding
    
    # Compare faces
    similarity_score, is_same_person = compare_faces(emb1, emb2)
    
    print(f"Similarity Score: {similarity_score:.4f}")
    print(f"Same person? {'YES' if is_same_person else 'NO'}")
    print(len(face1))
    # x1, y1, x2, y2 = face.bbox.astype(int)
    # cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)

    # cv2.imshow("Detected Face", img)
    # cv2.waitKey(0)
    # cv2.destroyAllWindows()

    # left_eye_color = get_eye_color(img, kps[0])
    # right_eye_color = get_eye_color(img, kps[1])
    
    # print("Left eye color:", left_eye_color)
    # print("Right eye color:", right_eye_color)
    
except Exception as e:
    print(f"Error: {str(e)}")
```
